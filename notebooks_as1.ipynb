```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)

max_count = counts.max()
most_frequent = values[counts == max_count]

print("Most frequent value:", most_frequent)
print("Indices:", np.where(np.isin(x, most_frequent))[0])

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)

max_count = counts.max()
most_frequent = values[counts == max_count]

print("Most frequent values:", most_frequent)
print("Indices:", np.where(np.isin(y, most_frequent))[0])
```

```python
g = np.matrix([[4, 1, 9],
              [12, 3, 1],
              [4, 5, 6]])

print("Sum of all elements:", np.sum(g))
print("Row-wise sum:", np.sum(g, axis=1))
print("Column-wise sum:", np.sum(g, axis=0))
```

```python
n_array = np.array([[55, 25, 15],
                    [30, 44, 2],
                    [11, 45, 77]])

print("Sum of diagonal elements:", np.trace(n_array))

eigenvalues, eigenvectors = np.linalg.eig(n_array)

print("Eigenvalues:")
print(eigenvalues)

print("Eigenvectors:")
print(eigenvectors)

print("Inverse:")
print(np.linalg.inv(n_array))

print("Determinant:")
print(np.linalg.det(n_array))
```

```python
from itertools import product
x = np.array([[2, 3, 4],
              [3, 2, 9]])

y = np.array([[1, 5, 0],
              [5, 10, 3]])
print(np.inner(x, y))
print(np.outer(x, y))
cartesian = list(product(x.flatten(), y.flatten()))

print(cartesian)
```

```python
# Q4
import numpy as np
from PIL import Image
import os

def img_to_array(path):
    img = Image.open(path)

    if img.mode == "RGB":
        arr = np.array(img)
        output_path = os.path.splitext(path)[0] + "_rgb.txt"

        with open(output_path, "w") as f:
            f.write(f"# shape: {arr.shape[0]} {arr.shape[1]} {arr.shape[2]}\n")
            np.savetxt(f, arr.reshape(-1, 3), fmt="%d")

    else:
        img = img.convert("L")
        arr = np.array(img)
        output_path = os.path.splitext(path)[0] + "_gray.txt"

        with open(output_path, "w") as f:
            f.write(f"# shape: {arr.shape[0]} {arr.shape[1]}\n")
            np.savetxt(f, arr, fmt="%d")

    print("Saved:", output_path)
img_to_array("image.jpg")
img_to_array("gray_image.jpg")
```

```python
import numpy as np

def load_array(path):
    with open(path, "r") as f:
        header = f.readline().strip()

    shape = tuple(map(int, header.replace("# shape:", "").split()))

    arr = np.loadtxt(path)

    return arr.reshape(shape)
arr = load_array("image_rgb.txt")

print(arr.shape)
print(arr)
```
