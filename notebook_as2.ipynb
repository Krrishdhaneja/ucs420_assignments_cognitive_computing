```python
# Enter your own details here
roll_number = input('Enter your roll number: ').strip()
name = input('Enter your name: ').strip()
branch = input('Enter your branch: ').strip()
age = int(input('Enter your age: '))
city = input('Enter your home city: ').strip()
cgpa = float(input('Enter your CGPA: '))

if not roll_number.isdigit():
    raise ValueError('Roll number must contain digits only.')

digits = [int(d) for d in roll_number]

if len(digits) < 8:
    raise ValueError('The assignment requires at least 8 digits of the roll number.')

print('Roll number:', roll_number)
print('Digits:', digits)
```

```python
# Q1: Create L
L = [digit * 10 for digit in digits]
print('i. L =', L)
```

```python
# ii. append() adds an element at the end of the list.
L.append(500)
print('After append(500):', L)

# insert() adds an element at a specified index.
L.insert(2, 250)
print('After insert(2, 250):', L)
```

```python
# iii. remove() removes the first occurrence of a specified value.
L.remove(500)
print('After remove(500):', L)

# pop() removes and returns the element at the specified index.
removed_element = L.pop(2)
print('Popped element:', removed_element)
print('After pop(2):', L)
```

```python
# iv. Sort ascending, then descending
L.sort()
print('Ascending:', L)

L.sort(reverse=True)
print('Descending:', L)
```

```python
# v. Slicing
print('First three elements:', L[:3])
print('Last three elements:', L[-3:])
```

```python
# vi. Elements greater than the average of L
average_L = sum(L) / len(L)
greater_than_average = [x for x in L if x > average_L]
print('Average of L:', average_L)
print('Elements greater than average:', greater_than_average)
```

```python
# Use the first 8 values from the original multiplied-digit list.
scores = tuple(digit * 10 for digit in digits[:8])
print('scores =', scores)
```

```python
# i. Highest score and its index; lowest score and its frequency
highest_score = max(scores)
highest_index = scores.index(highest_score)
lowest_score = min(scores)
lowest_count = scores.count(lowest_score)

print('Highest score:', highest_score)
print('Index of highest score:', highest_index)
print('Lowest score:', lowest_score)
print('Number of times lowest score appears:', lowest_count)
```

```python
# ii. A tuple is immutable, so it cannot be reversed in place.
reversed_scores = list(reversed(scores))
print('Reversed tuple returned as a list:', reversed_scores)
```

```python
# iii. Search for a user-entered score
score_input = int(input('Enter a score to search in scores: '))
if score_input in scores:
    print('First occurrence index:', scores.index(score_input))
else:
    print('Score is not present in the tuple.')
```

```python
# iv. Demonstrate tuple immutability.
# Running this assignment raises TypeError because tuple elements cannot be changed after creation.
try:
    scores[0] = 100
except TypeError as error:
    print('Error raised:', error)
    print('Reason: tuples are immutable, whereas list elements can be modified.')
```

```python
# v. Tuple unpacking with *
first_score, second_score, *remaining_scores = scores
print('First score:', first_score)
print('Second score:', second_score)
print('Remaining scores:', remaining_scores)
```

```python
import random

random.seed(int(roll_number))
random_numbers = [random.randint(100, 900) for _ in range(100)]
print('Generated list of 100 random numbers:')
print(random_numbers)
```

```python
# ii. Count and print odd numbers
odd_numbers = [x for x in random_numbers if x % 2 != 0]
print('Odd numbers count:', len(odd_numbers))
print('Odd numbers:', odd_numbers)
```

```python
# iii. Count and print even numbers
even_numbers = [x for x in random_numbers if x % 2 == 0]
print('Even numbers count:', len(even_numbers))
print('Even numbers:', even_numbers)
```

```python
# iv. Count and list prime numbers
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_numbers = [x for x in random_numbers if is_prime(x)]
print('Prime numbers count:', len(prime_numbers))
print('Prime numbers:', prime_numbers)
```

```python
# v. Most frequently occurring number and its frequency
frequency = {x: random_numbers.count(x) for x in set(random_numbers)}
most_frequent_number = max(frequency, key=frequency.get)
most_frequent_count = frequency[most_frequent_number]
print('Most frequent number:', most_frequent_number)
print('Frequency:', most_frequent_count)
```

```python
first_8_digits = digits[:8]
A = {digit * 7 for digit in first_8_digits}
B = {digit * 9 for digit in first_8_digits}

print('A =', A)
print('B =', B)
```

```python
# vi. Union
print('A union B =', A.union(B))

# vii. Intersection
print('A intersection B =', A.intersection(B))

# viii. Differences
print('A - B =', A.difference(B))
print('B - A =', B.difference(A))
print('difference() gives elements in one set but not the other; symmetric_difference() gives elements in either set but not both.')

# ix. Symmetric difference
print('Symmetric difference =', A.symmetric_difference(B))

# x. Subset and superset checks
print('A is subset of B:', A.issubset(B))
print('B is superset of A:', B.issuperset(A))
```

```python
# xi. discard() removes X if present and does nothing if it is absent.
X = int(input('Enter a value X to discard from A: '))
A.discard(X)
print('A after discard(X):', A)
print('discard() is safer than remove() because discard() does not raise an error when X is absent.')
```

```python
my_dict = {
    'name': name,
    'roll_no': roll_number,
    'branch': branch,
    'age': age,
    'city': city
}

print('Original dictionary:')
print(my_dict)
```

```python
# i. Rename city -> location using pop(), without recreating the dictionary by hand.
my_dict['location'] = my_dict.pop('city')
print('After renaming city to location:', my_dict)

# ii. Add cgpa
my_dict['cgpa'] = cgpa
print('After adding cgpa:', my_dict)

# iii. Increase age by 1
my_dict['age'] += 1
print('After increasing age by 1:', my_dict)
```

```python
# iv. Delete branch using pop() in one copy and del in another.
dict_pop = my_dict.copy()
removed_branch = dict_pop.pop('branch')

dict_del = my_dict.copy()
del dict_del['branch']

print('Using pop():', dict_pop)
print('Value returned by pop():', removed_branch)
print('Using del:', dict_del)
print('pop() returns the removed value, while del removes the key without returning its value.')
```

```python
# v. Iterate using items()
for key, value in my_dict.items():
    print(f'{key} → {value}')
```

```python
# vi. Safely check whether email exists before accessing it.
if 'email' in my_dict:
    print('Email:', my_dict['email'])
else:
    print('Email key does not exist.')
```

```python
# vii. Create a fictional friend's dictionary and merge it with our dictionary.
friend_dict = {
    'name': 'Rahul',
    'roll_no': 'FRIEND001',
    'branch': 'CSE',
    'age': 20,
    'city': 'Delhi'
}

merged_dict = {**my_dict, **friend_dict}
print('Merged dictionary:', merged_dict)
print('When both dictionaries contain the same key, the value from the later dictionary wins.')
```

```python
# viii. Keep only key-value pairs whose values are strings.
string_values_dict = {key: value for key, value in my_dict.items() if isinstance(value, str)}
print('Dictionary containing only string values:', string_values_dict)
```
